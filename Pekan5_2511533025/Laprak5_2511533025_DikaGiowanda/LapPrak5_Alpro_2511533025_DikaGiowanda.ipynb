## 2.1 Dasar Teori
Struktur kontrol perulangan (looping) merupakan mekanisme dasar dalam pemrograman yang digunakan untuk mengulang eksekusi suatu blok kode program secara otomatis selama kondisi tertentu terpenuhi. Penggunaan perulangan bertujuan untuk mengoptimalkan efisiensi kode, menghindari duplikasi baris kode, dan mempermudah pemrosesan data berjumlah besar.
Berikut adalah beberapa konsep utama yang mendasari perulangan for dan nested for:
1. Konsep Dasar Perulangan For

    Perulangan for adalah jenis struktur kontrol pengulangan yang terhitung. Struktur ini biasa digunakan ketika jumlah iterasi atau pengulangan kode sudah dapat ditentukan atau diketahui sebelum blok kode dijalankan. Perulangan for memiliki tiga elemen kontrol utama yang berada dalam satu baris deklarasi:

    a. Inisialisasi (Initialization): Penentuan nilai awal untuk variabel penyimpan iterasi (penghitung/counter).

    b. Kondisi (Condition/Terminating Condition): Ekspresi logika atau boolean yang diperiksa sebelum setiap iterasi dijalankan. Jika bernilai true, blok kode akan dieksekusi; jika false, perulangan akan berhenti.
    
    c. Pengubah Nilai (Increment/Decrement): Instruksi untuk mengubah (menambah atau mengurangi) nilai variabel kontrol setelah tiap iterasi selesai dieksekusi.

2. Konsep Perulangan Bersarang (Nested For)

    Nested for adalah struktur di mana terdapat satu atau lebih perulangan for di dalam blok perulangan for lainnya. Dalam skenario ini:

    a. Perulangan Luar (Outer Loop): Mengontrol iterasi tingkat pertama (misalnya baris).

    b. Perulangan Dalam (Inner Loop): Dieksekusi secara penuh dari awal hingga akhir pada setiap satu kali iterasi perulangan luar (misalnya kolom).

    c. Total jumlah iterasi keseluruhan yang terjadi adalah hasil perkalian antara jumlah iterasi perulangan luar dengan jumlah iterasi perulangan dalam.

3. Komponen Pengontrol Alur Perulangan (Loop Control Statements)

    Dalam eksekusi perulangan for, terdapat beberapa perintah khusus yang dapat digunakan untuk mengendalikan alur iterasi secara dinamis:

    a. Perintah break: Digunakan untuk menghentikan seluruh proses perulangan secara paksa dan keluar dari blok perulangan sebelum kondisi utama menjadi false.

    b. Perintah continue: Digunakan untuk mengabaikan atau melompati sisa instruksi di dalam blok perulangan pada iterasi saat ini dan langsung melanjutkan ke iterasi berikutnya.

4. Penerapan dan Penggunaan Kasus nyata (Use Cases)

    a. Perulangan For Tunggal: Sangat cocok digunakan untuk mengakses barisan data linier seperti elemen array satu dimensi, mencetak deret angka (aritmatika), serta melakukan kalkulasi akumulasi data (misalnya penjumlahan total nilai).

    b. Perulangan Nested For: Sangat ideal digunakan untuk mengolah data multidimensi, seperti operasi matematika matriks (2D/3D), traversal elemen array dua dimensi, pembuatan pola karakter visual (seperti segitiga atau piramida bintang), serta algoritma pengurutan dasar (seperti Bubble Sort dan Selection Sort).

### 2.2.1 Program "perulangan_for1_2511533025"

```python
# Buat file dengan nama perulangan_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_3025 = int(input("Masukan jumlah perulangan: "))
for i_3025 in range(ulang_3025):
    print(f"Perulangan ke-{i_3025}")
```

2. Bagian Kedua (for i in range(ulang):)

    Berfungsi sebagai deklarasi struktur perulangan for. Fungsi range(ulang) menghasilkan urutan angka yang dimulai dari 0 hingga ulang - 1. Pada setiap pengulangan, nilai urutan tersebut disimpannya secara bertahap ke dalam variabel iterasi i.

3. Bagian Ketiga (print(f"Perulangan ke-{i}"))

    Berfungsi untuk menampilkan baris teks keluaran ke layar pada setiap iterasi. Penggunaan huruf f sebelum tanda petik menandakan formatted string literal (f-string), yang memungkinkan nilai dari variabel i dimasukkan secara langsung ke dalam string menggunakan kurung kurawal {i}.

**Penjelasan:**

    d. Baris Baru (print()): Berfungsi untuk mencetak baris baru (pindah baris) agar tampilan hasil perulangan berikutnya berada di bawah baris perulangan pertama.

    b. Struktur Perulangan (for i in range(1, ulang+1):): Menggunakan fungsi range(star, stop) dengan batas awal 1 dan batas akhir ulang + 1. Dalam Python, nilai pada variabel stop bersifat exclusive (tidak diikutkan), sehingga perulangan berhenti tepat pada nilai ulang.

    c. Cetak Elemen (print(i, end=" ")): Mencetak nilai i dari 1 hingga ulang secara mendatar dengan pemisah spasi.

### 2.2.3 Program "perulangan_for3_2511533025"

```python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_3025 = int(input("Masukan jumlah perulangan: "))

jumlah_3025 = 0
for i_3025 in range(1, ulang_3025 + 1):
    print(i_3025, end=" ")
    jumlah_3025 = jumlah_3025 + i_3025

    if i_3025 < ulang_3025:
        print(" + ", end="")
    else:
        print(" = ", jumlah_3025, end="")
print()
print("Jumlah =", jumlah_3025)
```

**Penjelasan:**
1. Bagian Input Data (ulang = int(input("Masukan jumlah perulangan: ")))

    Berfungsi untuk mengambil nilai masukan angka dari pengguna. Teks masukan dikonversi dari tipe data string menjadi integer menggunakan fungsi int() dan disimpan dalam variabel ulang sebagai penentu batas akhir deret angka.

2. Inisialisasi Variabel Akumulator (jumlah = 0)

    Berfungsi untuk membuat variabel penampung total jumlah dengan nilai awal 0. Variabel jumlah nantinya digunakan untuk menjumlahkan setiap nilai iterasi yang muncul selama perulangan.

3. Struktur Perulangan for (for i in range(1, ulang + 1):)

    Menggunakan rentang dari 1 hingga ulang. Penggunaan ulang + 1 memastikan bahwa nilai maksimal ulang ikut diproses dalam perulangan.

4. Pencetakan Angka dan Akumulasi Penjumlahan

    a.print(i, end=" "): Mencetak nilai variabel perulangan i secara mendatar diikuti oleh spasi.

    b. jumlah = jumlah + i: Menambahkan nilai iterasi i ke dalam variabel akumulator jumlah secara bertahap pada setiap perulangan.

5. Pengondisian Karakter Pemisah / Operator (if-else)

    a. if i < ulang: Memeriksa apakah nilai iterasi saat ini masih kurang dari nilai batas akhir. Jika true, program mencetak simbol tambah " + " agar memisahkan antar-angka.

    b. else: Dipanggil ketika perulangan mencapai angka terakhir (i == ulang). Program mencetak simbol sama dengan dan hasil penjumlahan (" = ", jumlah) di ujung deret.

6. Baris Penutup dan Cetak Hasil Total

    a. print(): Berfungsi untuk melakukan ganti baris (pindah ke baris baru).

    b. print("Jumlah =", jumlah): Mencetak kembali hasil total penjumlahan pada baris baru di bawah ekspresi matematika.

### 2.2.4 Program "nested_for1_2511533025"

```python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_3025 = int(input("Masukkan nilai batas: "))
for line_3025 in range(1, batas_3025 + 1):
    for j_3025 in range(1, (-1 * line_3025 + batas_3025) + 1):
        print(".", end="")
    print(line_3025)
```

**Penjelasan:**
1. Input Batas Perulangan (batas = int(input("Masukkan nilai batas: ")))

    Berfungsi untuk menerima input angka dari pengguna, mengonversinya menjadi tipe data integer (int), lalu menyimpannya dalam variabel batas. Nilai ini menentukan berapa jumlah baris pola yang akan dicetak.

2. Perulangan Luar / Outer Loop (for line in range(1, batas + 1):)

    Berfungsi untuk mengontrol jumlah baris (vertikal) yang dicetak, dimulai dari baris ke-1 hingga baris ke-batas. Variabel line bertindak sebagai penanda nomor baris saat ini.   

3. Perulangan Dalam / Inner Loop (for j in range(1, (-1 * line + batas) + 1):)

    a. Berfungsi untuk mencetak karakter titik (.) secara horizontal pada baris yang sedang diproses.

    b. Formula matematika batas akhirnya adalah (-1 * line + batas) atau sederhananya (batas_3025 - line).
    
    c. Ditambah + 1 agar batas akhir nilai tersebut diikutsertakan dalam range(). Artinya, jumlah titik yang dicetak pada tiap baris akan berkurang 1 seiring bertambahnya nomor baris.

4. Pencetakan Karakter Titik (print(".", end=""))

    Berfungsi untuk mencetak satu karakter titik tanpa membuat baris baru. Penggunaan end="" menjaga agar titik-titik selanjutnya pada inner loop dicetak berdampingan secara mendatar.

5. Pencetakan Angka Baris dan Ganti Baris (print(line))

    Dieksekusi setelah inner loop selesai mencetak seluruh titik pada baris tersebut. Fungsi ini mencetak nomor baris (line) tepat di sebelah titik terakhir, lalu secara otomatis melakukan ganti baris (newline) untuk berpindah ke baris berikutnya.

### 2.2.5 Program "nested_for2_2511533025"

```python
# Buat file dengan nama nested_for2_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_3025 = int(input("Masukkan nilai batas: "))
for i_3025 in range(1, batas_3025 + 1):
    for j_3025 in range(1, batas_3025 + 1):
        print("*", end="")
    print()  # pindah ke baris berikutnya
```

**Penjelasan:**
1. Input Nilai Batas (batas = int(input("Masukkan nilai batas: ")))

    Berfungsi untuk menerima masukan berupa angka dari pengguna melalui fungsi input(). Fungsi int() mengubah masukan bernilai string menjadi tipe data integer (bilangan bulat). Nilai ini disimpan dalam variabel batas yang menjadi penentu dimensi (panjang dan lebar) matriks/persegi bintang yang akan terbentuk.   

2. Perulangan Luar / Outer Loop (for i in range(1, batas + 1):)

    a. Berfungsi untuk mengontrol jumlah baris secara vertikal.

    b. Fungsi range(1, batas + 1) membuat iterasi berjalan mulai dari angka 1 hingga tepat mencapai nilai batas.

    c. Setiap satu kali iterasi pada perulangan luar menandakan proses pembuatan satu baris baru.

3. Perulangan Dalam / Inner Loop (for j in range(1, batas + 1):)

    a. Berfungsi untuk mengontrol jumlah kolom secara horizontal pada baris yang sedang aktif.

    b. Sama seperti outer loop, perulangan dalam ini juga dieksekusi dari 1 hingga batas. Pada setiap iterasinya, perulangan dalam bertugas mencetak karakter bintang (*).

4. Pencetakan Karakter Bintang (print("*", end=""))

    Berfungsi untuk mencetak satu buah karakter bintang. Penggunaan parameter end="" bertujuan untuk mencegah fungsi print() membuat baris baru, sehingga bintang-bintang berikutnya pada baris yang sama akan dicetak bersampingan secara horizontal.

5. Pindah Baris (print() # pindah ke baris berikutnya)

    Dieksekusi setelah perulangan dalam (inner loop) selesai menjalankan seluruh iterasinya. Perintah print() tanpa argumen ini berfungsi untuk mencetak karakter newline (ganti baris), sehingga cetakan bintang pada iterasi perulangan luar berikutnya berpindah ke baris di bawahnya.

### 2.2.6 Program "nested_for3_2511533025"

```python
# Buat file dengan nama nested_for3_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_3025 = int(input("Masukan nilai batas: "))
for i_3025 in range(batas_3025+1):
    for j_3025 in range(batas_3025+1):
        print(i_3025+j_3025, end=" ")
    print() # pindah ke baris berikutnya
```

    Berfungsi untuk mengontrol nilai baris (indeks vertikal). Karena fungsi range(batas+1) hanya menyertakan satu parameter, perulangan dimulai secara otomatis dari angka 0 sampai dengan batas. Variabel i bertindak sebagai penampung angka indeks baris saat ini.

3. Perulangan Dalam / Inner Loop (for j in range(batas+1):)

    Berfungsi untuk mengontrol nilai kolom pada baris yang sedang aktif. Sama seperti perulangan luar, perulangan dalam ini berjalan dari angka 0 hingga batas. Variabel j bertindak sebagai penampung angka indeks kolom saat ini.

4. Pencetakan Hasil Penjumlahan (print(i+j, end=" "))

    Berfungsi untuk menghitung nilai hasil penjumlahan antara variabel baris (i) dan variabel kolom (j), lalu mencetaknya ke layar. Penggunaan parameter end=" " memastikan bahwa hasil cetakan dicetak mendatar ke samping dengan dipisahkan spasi.

5. Pindah Baris (print() # pindah ke baris berikutnya)

    Berfungsi untuk melakukan ganti baris setelah perulangan dalam selesai mengeksekusi semua kolom pada baris tersebut. Hal ini menyebabkan iterasi perulangan luar berikutnya dicetak di baris baru di bawahnya.

### 2.2.7 Program "nested_for4_2511533025"

```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_3025 = int(input("Masukan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3025 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3025 = tinggi_3025
    c_3025 = a_3025
    lebar_3025 = (2 * tinggi_3025) - 2

    for i_3025 in range(1, tinggi_3025 + 1):
        b_3025 = c_3025 + 1
        for j_3025 in range(1, lebar_3025 + 1):

            # Baris atas dan bawah
            if i_3025 == 1 or i_3025 == tinggi_3025:
                if j_3025 == 1 or j_3025 == lebar_3025:
                    print("#", end="")
                else:
                    print("=", end="")
                    
            # Baris isi
            else:
                if j_3025 == 1 or j_3025 == lebar_3025:
                    print("|", end="")
                else:
                    if j_3025 == c_3025:
                        print("<", end="")
                    elif j_3025 == b_3025:
                        print(">", end="")
                    elif j_3025 == (lebar_3025 - c_3025):
                        print("<", end="")
                    elif j_3025 == (lebar_3025 - c_3025 + 1):
                        print(">", end="")
                    elif j_3025 > b_3025 and j_3025 < (lebar_3025 - c_3025):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_3025 -= 2

        if a_3025 <= 0:
            c_3025 = (-a_3025) + 2
        else:
            c_3025 = a_3025
```

**Penjelasan:**
1. Input dan Validasi Bilangan Genap

    a. tinggi = int(input(...)): Mengambil masukan dari pengguna dan mengonversinya menjadi tipe data integer. Nilai ini menentukan tinggi dari pola bingkai/pola grafis yang dibentuk.

    b. if tinggi % 2 != 0:: Memeriksa apakah input bernilai ganjil menggunakan operator sisa bagi (%). Jika input ganjil, program mencetak peringatan "Tinggi harus bilangan genap!" dan proses pembuatan pola dibatalkan. Hal ini dilakukan karena simetri bagian dalam pola membutuhkan jumlah baris genap.

2. Inisialisasi Variabel Penentu Dimensi dan Offset

    a. a_3025 = tinggi: Variabel pembantu untuk mengontrol penggeseran posisi tanda panah pada bagian dalam bingkai.

    b. c = a: Menimpan offset posisi kolom horizontal untuk penempatan karakter <.

    c. lebar = (2 * tinggi) - 2: Menghitung lebar total bingkai. Lebar dibuat proporsional terhadap tinggi program.

3. Perulangan Luar / Outer Loop (for i in range(1, tinggi + 1):)

    Berfungsi untuk mengontrol cetakan dari baris pertama (i = 1) hingga baris terakhir (i = tinggi).

4. Perulangan Dalam / Inner Loop (for j in range(1, lebar + 1):)

    Mengontrol pencetakan karakter per kolom dari kiri ke kanan pada baris yang sedang aktif.

5. Logika Percabangan Pembentukan Pola Visual (if-else):
    1. Batas Baris Atas dan Bawah (if i == 1 or i == tinggi:):

        a. Pada sudut paling kiri (j == 1) dan sudut paling kanan (j == lebar), program mencetak karakter #.
        
        b. Pada kolom di antara kedua sudut tersebut, program mencetak garis horizontal =.

    2. Baris Isi / Tengah (Selain Baris Pertama dan Terakhir):

        a. Batas Dinding Kiri dan Kanan: Jika j == 1 atau j == lebar, program mencetak garis vertikal |.

        b. Penempatan Karakter Panah Kiri (<): Dicetak saat kolom j sama dengan c atau (lebar - c).

        c. Penempatan Karakter Panah Kanan (>): Dicetak saat kolom j sama dengan b atau (lebar - c + 1).

        d. Isi Tengah Titik (.): Jika posisi kolom berada di antara dua pasang panah (antara b dan lebar - c), dicetak karakter titik ..
        
        e. Spasi Kosong ( ): Jika tidak memenuhi kondisi di atas, dicetak spasi kosong untuk membentuk ruang transisi.

6. Penyesuaian Offset Pasca Iterasi Baris (a -= 2 & Pembaharuan c)

    a. Setelah satu baris selesai dicetak, perintah print() dipanggil untuk melakukan ganti baris. Selanjutnya, nilai a dikurangi 2 pada tiap baris.
    
    b. Pada setengah bagian atas, nilai a bernilai positif sehingga c mengecil dan pasangan panah bergerak mendekati bagian tengah.

    c. Ketika a <= 0 (memasuki setengah bagian bawah), nilai c dihitung dengan rumus (-a) + 2, yang menyebabkan pasangan panah kembali bergerak menjauh ke arah luar, membentuk simetri cermin jam pasir.

### 2.2.8 Program "jumlah_genap_2511533025"

```python
# Buat file dengan nama jumlah_genap_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_3025 = int(input("Masukan nilai batas: "))

jumlah_3025 = 0
for i_3025 in range(1, ulang_3025 + 1):
    if i_3025 % 2 == 0:
        print(i_3025, end=" ")
        jumlah_3025 = jumlah_3025 + i_3025
        if i_3025 < ulang_3025:
            print(" + ", end="")
        else:
            print(" = ", jumlah_3025, end="")
print()
print("Jumlah =", jumlah_3025)
```

**Penjelasan:**
1. Input Nilai Batas (ulang = int(input("Masukan nilai batas: ")))

    Berfungsi untuk menerima masukan angka dari pengguna melalui fungsi input(). Nilai masukan yang awalnya bertipe data string dikonversi menjadi tipe data integer menggunakan int() dan disimpan dalam variabel ulang. Nilai ini berfungsi sebagai batas atas deret angka yang akan diperiksa.

2. Inisialisasi Variabel Akumulator (jumlah = 0)

    Berfungsi untuk membuat variabel penampung total hasil penjumlahan yang diisi dengan nilai awal 0. Variabel ini bertugas menjumlahkan angka-angka genap yang ditemukan selama perulangan berjalan.

3. Struktur Perulangan for (for i in range(1, ulang + 1):)

    Berfungsi untuk melakukan iterasi angka secara berurutan mulai dari 1 hingga ulang. Penambahan + 1 pada fungsi range() memastikan bahwa nilai ulang sendiri tetap diikutsertakan dalam pemeriksaan.

4. Pemeriksaan Bilangan Genap (if i % 2 == 0:)

    a. Menggunakan operator modulo (%) untuk memeriksa apakah nilai i habis dibagi 2 (sisa bagi bernilai 0).

    b. Jika ya (bilangan genap), maka blok kode di dalam if akan dieksekusi.

    c. Jika tidak (bilangan ganjil), nilai tersebut diabaikan dan perulangan berlanjut ke angka berikutnya.

5. Pencetakan Angka dan Akumulasi Penjumlahan

    a. print(i, end=" "): Mencetak angka genap yang ditemukan secara mendatar ke samping dengan dipisahkan spasi.

    b. jumlah = jumlah + i: Menambahkan nilai angka genap tersebut ke dalam variabel jumlah secara bertahap.

6. Pengondisian Tanda Pemisah dan Sama Dengan (if-else Bersarang)

    a. if i < ulang: Memeriksa apakah angka genap saat ini masih belum mencapai batas nilai maksimal ulang. Jika bernilai true, program mencetak operator tambah " + ".

    b. else: Dipanggil ketika angka genap yang diproses merupakan angka terakhir pada batas rentang (i == ulang atau angka genap terbesar yang mendekati batas). Program mencetak tanda sama dengan beserta hasil akumulasi (" = ", jumlah).

7. Ganti Baris dan Cetak Total Keseluruhan

    a. print(): Berfungsi untuk melakukan ganti baris (newline).

    b. print("Jumlah =", jumlah): Mencetak ulang hasil total akumulasi bilangan genap pada baris baru di bawah ekspresi matematika.

```python
# Tugas Pekan 5 - Pola Jam Pasir Kristal Palindromik Berbingkai
# Tambahkan varaibel 4 digit terakhir NIM.

print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK ===")
n_3025 = int(input("Masukkan ukuran skala jam pasir (N): "))

# ---------- Bingkai atas ----------
print("#", end="")
for garis_3025 in range(4 * n_3025 + 5):
    print("=", end="")
print("#", end="")
print()

# ---------- Fase 1: Jam pasir atas (N turun s.d. 1) ----------
for baris_3025 in range(n_3025, 0, -1):
    print("| ", end="")
    for spasi_3025 in range(2 * (n_3025 - baris_3025)):
        print(" ", end="")
    for angka_3025 in range(baris_3025, 0, -1):
        print(angka_3025, end=" ")
    print("<*>", end="")
    for angka_3025 in range(1, baris_3025 + 1):
        print(" ", end="")
        print(angka_3025, end="")
    for spasi_3025 in range(2 * (n_3025 - baris_3025)):
        print(" ", end="")
    print(" |", end="")
    print()

# ---------- Fase 2: Poros titik pusat ----------
print("|", end="")
for spasi_3025 in range(2 * n_3025 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_3025 in range(2 * n_3025 + 1):
    print(" ", end="")
print("|", end="")
print()

# ---------- Fase 3: Jam pasir bawah (1 naik s.d. N) ----------
for baris_3025 in range(1, n_3025 + 1):
    print("| ", end="")
    for spasi_3025 in range(2 * (n_3025 - baris_3025)):
        print(" ", end="")
    for angka_3025 in range(baris_3025, 0, -1):
        print(angka_3025, end=" ")
    print("<*>", end="")
    for angka_3025 in range(1, baris_3025 + 1):
        print(" ", end="")
        print(angka_3025, end="")
    for spasi_3025 in range(2 * (n_3025 - baris_3025)):
        print(" ", end="")
    print(" |", end="")
    print()

# ---------- Bingkai bawah ----------
print("#", end="")
for garis_3025 in range(4 * n_3025 + 5):
    print("=", end="")
print("#", end="")
print()
```

    Menggunakan loop yang berjalan mundur dari N turun hingga 1. Pada setiap baris, program mencetak elemen-elemen berikut secara berurutan:
