## 2.1 Dasar Teori
Dalam bidang algoritma dan pemrograman, pemrosesan data oleh komputer membutuhkan penyimpanan sementara dan aturan mengenai jenis data yang diolah. Konsep mendasar yang mengatur hal tersebut adalah variabel dan tipe data.
1. Pengertian Variabel

    Variabel adalah sebuah nama atau simbol yang digunakan untuk mengidentifikasi suatu lokasi penampungan data di dalam memori komputer (RAM). Nilai yang tersimpan dalam variabel dapat diakses, diubah, atau dimanipulasi selama program berlangsung.

2. Aturan Penamaan Variabel (Naming Conventions)

    Dalam menuliskan nama variabel, terdapat beberapa aturan baku yang harus dipatuhi agar tidak terjadi syntax error:

    a. Nama variabel bersifat case sensitive (membedakan huruf besar dan huruf kecil, contoh: nilai berbeda dengan Nilai).

    b. Harus diawali dengan huruf atau karakter underscore (_), tidak boleh diawali dengan angka.

    c. Tidak boleh menggunakan kata kunci khusus dalam bahasa pemrograman (seperti if, for, class, import).

    d. Tidak boleh memuat spasi atau karakter khusus/simbol seperti @, #, $, %.

3. Deklarasi dan Inisialisasi Variabel

    a. Deklarasi: Proses mendaftarkan nama variabel beserta tipe datanya ke sistem agar memori dialokasikan.

    b. Inisialisasi: Proses pemberian nilai awal (initial value) pada variabel yang telah dideklarasikan.

4. Pengertian Tipe Data

    Tipe Data adalah klasifikasi yang memberitahu sistem atau interpreter mengenai jenis nilai yang dikandung oleh suatu variabel, bagaimana data tersebut disimpan di memori, serta operasi aritmatika atau logika apa saja yang dapat dilakukan terhadap data tersebut.

5. Klasifikasi Tipe Data Dasar (Primitive Data Types)

    a. Integer (Bilangan Bulat): Tipe data yang digunakan untuk menyimpan angka bulat positif maupun negatif tanpa koma/desimal (contoh: -5, 0, 42).

    b. Float / Double (Bilangan Desimal): Tipe data yang menyimpan angka pecahan atau memiliki koma desimal (contoh: 3.14, -0.001, 2.5).

    c. String (Teks / Karakter Kombinasi): Tipe data yang menyimpan sekumpulan karakter berupa teks, angka, maupun simbol yang diapit oleh tanda petik (contoh: "Hello World", "A123").

    d. Boolean (Nilai Logika): Tipe data sederhana yang hanya memiliki dua kemungkinan nilai, yaitu True (Benar) atau False (Salah), yang umumnya digunakan dalam percabangan logika.

    e. Character (Char): Tipe data yang digunakan untuk menyimpan tunggal satu karakter (contoh: 'A', 'z', '8').

6. Konversi Tipe Data (Type Casting)
    Konversi tipe data adalah proses mengubah nilai dari satu tipe data ke tipe data lainnya. Terdiri dari dua jenis:

    a.Implicit Casting (Otomatis): Konversi yang dilakukan secara otomatis oleh bahasa pemrograman ketika mengubah dari tipe data berkapasitas kecil ke besar.

    b. Explicit Casting (Manual): Konversi yang dilakukan secara sengaja oleh programmer menggunakan fungsi khusus (contoh: mengubah String "123" menjadi Integer 123).

7. Konstanta
    Konstanta adalah jenis variabel khusus yang nilainya bersifat tetap dan tidak dapat diubah setelah didefinisikan selama program dijalankan (contoh: nilai Pi = 3.14159).

8. Cakupan Variabel (Variable Scope)
    
    a. Variabel Lokal: Variabel yang dideklarasikan di dalam suatu fungsi/blok tertentu dan hanya bisa diakses di dalam blok tersebut.

    b. Variabel Global: Variabel yang dideklarasikan di luar fungsi/blok sehingga dapat diakses oleh seluruh bagian program.

## 2.2 Kode Program

### 2.2.1 Program "Numerik_2511533025"


```python
# Buat file dengan nama Numerik_NIM.py
# Create a variable with integer value.
a_3025=100
print("The type of variable having value", a_3025, " is ", type(a_3025))

# Create a variable with float value.
b_3025=10.2345
print("The type of variable having value", b_3025, " is ", type(b_3025))

# Create a variable with complex value.
c_3025=100+3j
print("The type of variable having value", c_3025, " is ", type(c_3025))
```

**Penjelasan :**
1. Bagian pertama kode (Penggunaan Tipe Data Integer) berfungsi untuk menginisialisasi variabel a_3025 dengan nilai bilangan bulat 100. Fungsi type(a_3025) digunakan untuk mengecek kelas tipe data dari variabel tersebut. Hasil eksekusi menampilkan <class 'int'>, yang menunjukkan bahwa variabel a_3025 menyimpan data berjenis Integer (bilangan bulat tanpa koma/desimal).

2. Bagian kedua kode (Penggunaan Tipe Data Float) berfungsi untuk menginisialisasi variabel b_3025 dengan nilai desimal 10.2345. Ketika fungsi type(b_3025) dipanggil, program mengenali nilai ini sebagai bilangan pecahan/riil. Hasil eksekusi menampilkan <class 'float'>, yang menandakan bahwa variabel b_3025 memiliki tipe data Float (bilangan berkoma).

3. Bagian ketiga kode (Penggunaan Tipe Data Complex) berfungsi untuk mendefinisikan variabel c_3025 dengan nilai bilangan kompleks 100+3j, di mana 100 adalah bagian riil dan 3j adalah bagian imajiner (dalam Python digunakan huruf j). Hasil pemeriksaan fungsi type(c_3025) menghasilkan <class 'complex'>, yang mengonfirmasi bahwa variabel c_3025 merupakan tipe data Complex (bilangan kompleks).

4. Penggunaan Fungsi print() dan type() secara keseluruhan berfungsi untuk menampilkan output teks gabungan antara pesan deskriptif, nilai variabel yang tersimpan, serta tipe data resmi yang terdeteksi oleh interpreter Python secara dinamis (dynamic typing).

### 2.2.2 Program "Boolean_2511533025"

```python
# Buat file dengan nama Boolean_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasi variabel dengan tipe data Boolean
is_lulus_3025 = True
is_cumlaude_3025 = True

# Menggunakan Boolean
nilai_3025 = 85
batas_lulus_3025 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan_3025 = nilai_3025 >= batas_lulus_3025 # Hasilnya akan true

print("=== Check Kelulusan ===")
print("Nilai:", nilai_3025)
print("Apakah Lulus?:", status_kelulusan_3025)
if is_lulus_3025 and is_cumlaude_3025:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
```

**Penjelasan :**
1. Deklarasi Variabel Boolean secara Langsung

    Bagian ini mendefinisikan variabel is_lulus_3025 dan is_cumlaude_3025 yang bernilai True. Tipe data Boolean ini digunakan untuk menyimpan status kebenaran logika (True atau False).

2. Inisialisasi Variabel Numerik

    Bagian ini mendeklarasikan variabel nilai_3025 dengan nilai 85 dan batas_lulus_3025 dengan nilai 75. Kedua variabel ini memiliki tipe data Integer yang nantinya digunakan sebagai acuan perbandingan logika.

3. Penentuan Nilai Boolean dari Evaluasi Kondisi (Operator Komparasi)

    Variabel status_kelulusan_3025 menyimpan hasil perbandingan ekspresi logika nilai_3025 >= batas_lulus_3025. Karena nilai 85 lebih besar atau sama dengan 75, maka ini mengevaluasi ke nilai True, yang otomatis membuat status_kelulusan_3025 bertipe data Boolean dengan nilai True.

4. Menampilkan Informasi (Pencetakan Output)

    Fungsi print() digunakan untuk menampilkan teks, nilai dari variabel nilai_3025 (yaitu 85), serta hasil evaluasi logika pada status_kelulusan_3025 (yaitu True).

5. Pengujian Kondisi Logika Terstruktur (Operator Logika and)

    Struktur percabangan if mengevaluasi gabungan dua variabel Boolean menggunakan operator logika and (is_lulus_3025 and is_cumlaude_3025). Karena kedua variabel tersebut bernilai True, maka secara keseluruhan dan blok kode di dalamnya mengeksekusi perintah untuk mencetak pesan "Selamat, Anda lulus dengan predikat Cum Laude!".

### 2.2.3 Program "String_2511533025"

```python
# Buat file dengan nama String_NIM.py
# Nama variable ditambah 4 digit nim terakhir contoh: nama_1234
# Ganti programnya dan data masing2
nama_3025 = "Dika Giowanda"
jenis_kelamin_3025 = 'L'
alamat_3025 = """
    Padang Lawas
    kec. Luhak Nan Duo
    Kab. Pasaman Barat
"""
agama_3025 = 'islam'
print("Nama saya adalah: ", nama_3025)
print("Jenis kelamin saya adalah: ", jenis_kelamin_3025)
print("Alamat saya adalah: ", alamat_3025)
print("Agama saya adalah: ", agama_3025)
```

    Variabel nama_3025 diinisialisasi dengan nilai "Dika Giowanda". Penggunaan tanda petik ganda digunakan untuk menyimpan data berupa teks (string) dalam satu baris kalimat.

2. Inisialisasi Variabel String dengan Tanda Petik Tunggal (' ')

    Variabel jenis_kelamin_3025 bernilai 'L' dan agama_3025 bernilai 'islam'. Di Python, tanda petik tunggal memiliki fungsi yang sama dengan tanda petik ganda untuk mendefinisikan tipe data String, baik untuk satu karakter maupun teks berupa kata.

3. Deklarasi Multi-line String menggunakan Triple Quotes (""" """)
    
    Variabel alamat_3025 didefinisikan menggunakan tanda petik ganda tiga kali ("""). Fitur ini berfungsi untuk membuat teks string yang mencakup beberapa baris (multiline string), di mana format baris baru (newline) dan spasi di dalamnya tetap dipertahankan saat program dijalankan.

4. Menampilkan Data Teks (Output Formatting)
    
    Fungsi print() digunakan sebanyak empat kali untuk menampilkan label teks statis yang digabungkan dengan variabel teks. Penggunaan koma (,) di dalam fungsi print() berfungsi untuk menggabungkan string sekaligus otomatis menambahkan spasi di antara label teks dan isi variabel yang dipanggil.

### 2.2.4 Program "Input_2511533025"

```python
# Buat file dengan nama Input_2511533025.py
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: panjang_123417
# Program ini menggunakan fungsi input() untuk meminta pengguna
# Memasukan nilai panjang dan lebar persegi panjang.
# Nilai yang dimasukan akan dikonversi menjadi tipe data integer
panjang_3025 = input('Masukan nilai panjang: ')
lebar_3025 = input('Masukan nilai lebar: ')
luas_3025 = int(panjang_3025) * int(lebar_3025)
print("Luas =", luas_3025)
```

**Penjelasan :**
1. Menerima Input Dinamis dari Pengguna
    
    Fungsi input() pada variabel panjang_3025 dan lebar_3025 digunakan untuk menerima input dari pengguna melalui konsol. Secara standar (default), nilai yang ditangkap oleh fungsi input() di Python selalu bertipe data String (teks).

2. Konversi Tipe Data Eksplisit (Type Casting)
    
    Fungsi int() digunakan untuk melakukan type casting, yaitu mengubah tipe data masukan yang semula berbentuk String menjadi tipe data Integer. Hal ini wajib dilakukan karena operasi matematika tidak dapat langsung dilakukan pada tipe data String.

3. Operasi Aritmatika dan Perhitungan Luas
    
    Variabel luas_3025 menampung hasil perkalian antara int(panjang_3025) dan int(lebar_3025). Menggunakan operator perkalian (*), sistem menghitung luas persegi panjang berdasarkan nilai integer yang telah dikonversi (pada contoh output: 17×7=119).
    
4. Menampilkan Hasil Perhitungan (Output)

    Fungsi print("Luas =", luas_3025) digunakan untuk mencetak label teks "Luas =" beserta nilai akhir yang tersimpan dalam variabel luas_3025 ke layar terminal.

### 2.2.5 Program "Konstanta_2511533025"

```python
# Buat file dengan nama Konstanta_NIM.py
# Program ini menggunakan konstantas untuk menghitung luas lingkaran
# Nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_3025 = float(input('Masukan nilai jari-jari: '))
luas_3025 = PI * jari_3025 * jari_3025
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3025, luas_3025))
```

**Penjelasan :**
1. Import Modul dan Deklarasi Konstanta (Final)

    Perintah from typing import Final digunakan untuk mengambil penanda tipe Final. Variabel PI: Final = 3.14 mendefinisikan nilai konstanta Pi secara tetap.

2. Input dan Konversi Tipe Data Float

    Fungsi input() mengambil nilai jari-jari dari pengguna yang secara langsung dikonversi menggunakan float(...). Hal ini bertujuan agar variabel jari_3025 dapat menerima input berupa angka desimal maupun bilangan bulat untuk presisi perhitungan matematika.

3. Perhitungan Luas Lingkaran

    Variabel luas_3025 menghitung luas lingkaran berdasarkan rumus Luas=π×r2 dengan mengalikan konstanta PI dan nilai variabel jari_3025 secara berulang (PI * jari_3025 * jari_3025).

4. Formatting Output Menggunakan String Formatting (%)
    Penggunaan operator penanda format % digunakan untuk mengatur tampilan desimal pada output:
    
    a. %f menampilkan nilai float secara default dengan 6 digit desimal (misal: 3.140000).

    b. %.2f membatasi jumlah angka di belakang koma menjadi tepat 2 digit desimal saja (misal: 17.00 dan 907.46), sehingga tampilan variabel jari_3025 dan luas_3025 menjadi lebih rapi dan terbaca.
