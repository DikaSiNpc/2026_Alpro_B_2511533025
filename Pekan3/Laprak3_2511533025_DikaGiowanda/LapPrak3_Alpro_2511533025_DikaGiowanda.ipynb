## 2.1 Dasar Teori
Operator adalah simbol atau tanda dalam pemrograman yang digunakan untuk melakukan suatu operasi terhadap satu atau lebih nilai (operand), menghasilkan nilai baru sebagai keluarannya. Dalam algoritma dan pemrograman, operator berperan penting untuk melakukan perhitungan, pengambilan keputusan, manipulasi data, hingga pengujian kondisi logis. Berikut adalah jenis-jenis operator yang umum digunakan:
1. Operator Aritmatika
Digunakan untuk melakukan operasi perhitungan matematis dasar antara dua nilai numerik.

    a. + (Penjumlahan) — contoh: 5 + 3 menghasilkan 8.

    b. - (Pengurangan) — contoh: 5 - 3 menghasilkan 2.

    c. * (Perkalian) — contoh: 5 * 3 menghasilkan 15.

    d. / (Pembagian) — contoh: 5 / 2 menghasilkan 2.5.

    e. // (Pembagian bulat / floor division) — contoh: 5 // 2 menghasilkan 2.

    f. % (Modulus / sisa bagi) — contoh: 5 % 2 menghasilkan 1.

    g. ** (Perpangkatan) — contoh: 5 ** 2 menghasilkan 25.

2. Operator Logika
Digunakan untuk menggabungkan atau membalik nilai dari beberapa ekspresi kondisional (boolean), sering dipakai dalam struktur percabangan.

    a. and — bernilai benar jika kedua kondisi bernilai benar. Contoh: True and False menghasilkan False.

    b. or — bernilai benar jika salah satu kondisi bernilai benar. Contoh: True or False menghasilkan True.

    c. not — membalik nilai kebenaran suatu kondisi. Contoh: not True menghasilkan False.

3. Operator Perbandingan (Relasional)
Digunakan untuk membandingkan dua nilai dan menghasilkan nilai boolean (True/False).

    a. == (Sama dengan) — contoh: 5 == 5 menghasilkan True.
    
    b. != (Tidak sama dengan) — contoh: 5 != 3 menghasilkan True.
    
    c. > (Lebih besar dari) — contoh: 5 > 3 menghasilkan True.
    
    d. < (Lebih kecil dari) — contoh: 5 < 3 menghasilkan False.
    
    e. >= (Lebih besar atau sama dengan) — contoh: 5 >= 5 menghasilkan True.
    
    f. <= (Lebih kecil atau sama dengan) — contoh: 5 <= 3 menghasilkan False.

4. Operator Assignment (Penugasan)
Digunakan untuk memberikan atau memperbarui nilai suatu variabel. Selain operator = dasar, terdapat bentuk gabungan (compound assignment) yang mempersingkat penulisan.

    a. = (Penugasan nilai) — contoh: x = 5, artinya variabel x diberi nilai 5.
    
    b. += (Tambah lalu tugaskan) — contoh: x += 3 setara dengan x = x + 3.
    
    c. -= (Kurang lalu tugaskan) — contoh: x -= 3 setara dengan x = x - 3.
    
    d. *= (Kali lalu tugaskan) — contoh: x *= 3 setara dengan x = x * 3.
    
    e. /= (Bagi lalu tugaskan) — contoh: x /= 3 setara dengan x = x / 3.
    
    f. %= (Modulus lalu tugaskan) — contoh: x %= 3 setara dengan x = x % 3.
    
    g. **= (Pangkat lalu tugaskan) — contoh: x **= 2 setara dengan x = x ** 2.

5. Operator Bitwise
Digunakan untuk memanipulasi data pada level bit (representasi biner dari suatu bilangan).
    
    a. & (AND bitwise) — contoh: 5 & 3 menghasilkan 1.
    
    b. | (OR bitwise) — contoh: 5 | 3 menghasilkan 7.
    
    c. ^ (XOR bitwise) — contoh: 5 ^ 3 menghasilkan 6.
    
    d. ~ (NOT bitwise / komplemen) — contoh: ~5 menghasilkan -6.
    
    e. << (Geser bit ke kiri) — contoh: 5 << 1 menghasilkan 10.
    
    f. >> (Geser bit ke kanan) — contoh: 5 >> 1 menghasilkan 2.

6. Operator Lainnya: Keanggotaan (Membership) dan Identitas (Identity)
Operator keanggotaan digunakan untuk memeriksa apakah suatu nilai terdapat di dalam suatu struktur data seperti list, string, atau tuple:

    a. in — bernilai benar jika nilai tersebut ada di dalam struktur data. Contoh: 3 in [1, 2, 3] menghasilkan True.

    b. not in — bernilai benar jika nilai tersebut tidak ada di dalam struktur data. Contoh: 5 not in [1, 2, 3] menghasilkan True.

    Operator identitas digunakan untuk memeriksa apakah dua variabel merujuk ke objek yang sama persis di memori, bukan sekadar memiliki nilai yang sama:

    a. is — bernilai benar jika kedua variabel merujuk ke objek yang sama. Contoh: x is y.
    
    b. is not — bernilai benar jika kedua variabel merujuk ke objek yang berbeda. Contoh: x is not y.

### 2.2.1 Program "Aritmatika_2511533025"


```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam python
# Nama variable ditambah 4 digit nim terakhir contoh: nama_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukan akan dikonvesi menjadi data tipe integer

angka1_3025 = int(input("Input angka-1: "))
angka2_3025 = int(input("Input angka-2: "))

# Penjumlahan
hasil_3025 = angka1_3025 + angka2_3025
print("\nOperator Penjumlahan")
print("Hasil =", hasil_3025)

# Pengurangan
hasil_3025 = angka1_3025 - angka2_3025
print("\nOperator Pengurangan")
print("Hasil =", hasil_3025)

# Perkalian
hasil_3025 = angka1_3025 * angka2_3025
print("\nOperator Perkalian")
print("Hasil =", hasil_3025)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_3025 != 0:
    hasil_3025 = angka1_3025 / angka2_3025
    print("\nOperator Pembagian")
    print("Hasil =", hasil_3025)

    hasil_3025 = angka1_3025 // angka2_3025
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_3025)

    hasil_3025 = angka1_3025 % angka2_3025
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_3025)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_3025 = angka1_3025 ** angka2_3025
print("\nOperator Pangkat")
print("Hasil =", hasil_3025)
```

**Penjelasan :**
1. Inisialisasi Input: Menggunakan fungsi input() untuk menerima masukan angka dari pengguna, lalu dikonversi menjadi tipe data bulat menggunakan int() agar dapat digunakan dalam perhitungan matematis.   

2. Operator Penjumlahan (+): Menjumlahkan angka1 dan angka2.   

2. Operator Pengurangan (-): Mengurangi angka1 dengan angka2.   

3. Operator Perkalian (*): Mengalikan nilai kedua variabel.

4. Pengecekan Kondisi (if): Memastikan nilai angka2 tidak sama dengan 0 (!= 0) sebelum melakukan operasi pembagian untuk menghindari kesalahan division by zero.

5. Operator Pembagian (/): Membagi angka1 dengan angka2 yang menghasilkan nilai desimal (float).   

6. Operator Pembagian Bulat (//): Melakukan pembagian lalu membulatkan hasilnya ke bawah (floor division) menghasilkan bilangan bulat.   

7. Operator Sisa Bagi (%): Menghitung sisa hasil bagi (modulo) dari kedua angka.   

8. Operator Pangkat (**): Menghitung nilai angka1 dipangkatkan dengan angka2.

### 2.2.2 Program "Logika_2511533025"

```python
# Buat file dengan nama logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a1_1234
# Program ini menggunakan fungsi input()
# Program operator logika dalam Python

# Memasukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_3025 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_3025 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1 =", a1_3025)
print("A2 =", a2_3025)
# Konjungsi: bernilai True jika keduanya True
hasil_3025 = a1_3025 and a2_3025
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil_3025)

# Disjungsi: bernilai True jika salah satunya True
hasil_3025 = a1_3025 or a2_3025
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil_3025)

# Negasi A1: membalik nilai A1
hasil_3025 = not a1_3025
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil_3025)

# Negasi A2: membalik nilai A2
hasil_3025 = not a2_3025
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil_3025)

# XOR: bernilai True jika kedua nilai berbeda
hasil_3025 = a1_3025 != a2_3025
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil_3025)
```

**Penjelasan:**
1. Penerimaan Input Nilai Boolean secara Dinamis dan Pembersihan Teks (String Sanitization)
Baris input(...).strip().lower() == "true" digunakan untuk meminta masukan teks dari pengguna melalui terminal:

    a. Metode .strip() berfungsi untuk Disjungsipus spasi yang tidak sengaja terketik di awal atau akhir teks.

    b. Metode .lower() mengubah seluruh huruf masukan menjadi huruf kecil (lowercase), membuat program tidak peka terhadap huruf besar/kecil (case-insensitive) saat pengguna mengetik "True", "TRUE", atau "true".

    c. Perbandingan == "true" mengevaluasi teks tersebut dan mengembalikan nilai Boolean (True atau False) yang disimpan ke dalam variabel a1 dan a2.

2. Operasi Konjungsi (Operator Logika and)
Variabel hasil = a1 and a2 menjalankan operasi logika AND. Operasi ini hanya akan menghasilkan nilai True jika kedua operan bernilai True secara bersamaan. Jika salah satu atau keduanya bernilai False, maka hasilnya akan False.

3. Operasi Disjungsi (Operator Logika or)
Variabel hasil = a1 or a2 menjalankan operasi logika OR. Operasi ini akan menghasilkan nilai True apabila minimal salah satu dari variabel masukan bernilai True. Hasilnya hanya akan bernilai False jika kedua variabel bernilai False.

4. Operasi Negasi (Operator Logika not)
Baris hasil = not a1 dan not a2 menerapkan operator NOT untuk membalikkan nilai kebenaran (inversion). Jika nilai awal variabel adalah True, maka negasinya menjadi False, begitu pula sebaliknya.

5. Operasi Disjungsi Eksklusif (Operator Logika XOR)
Perintah hasil = a1 != a2 mensimulasikan gerbang logika XOR menggunakan operator ketidaksamaan (!=). Operasi ini menghasilkan nilai True jika nilai kedua variabel berbeda (satu True dan satu lagi False), dan bernilai False jika kedua nilai variabel tersebut sama.

6. Pencetakan Hasil Output
Fungsi print() digunakan secara berurutan untuk menampilkan status nilai awal variabel serta hasil dari setiap pengujian logika (Konjungsi, Disjungsi, Negasi, dan XOR) ke layar konsol secara terstruktur.

### 2.2.4 Program "Assignmnet_2511533025"

```python
# Buat file dengan nama assignment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam Python

angka1_3025 = int(input("Input angka-1: "))
angka2_3025 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1_3025)
print("Nilai angka2 =", angka2_3025)
# Assignment biasa
hasil_3025 = angka1_3025
print("\nAssignment biasa (=)")
print("Hasil =", hasil_3025)
# Assignment penambahan
hasil_3025 = angka1_3025
hasil_3025 += angka2_3025
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil_3025)

# Assignment pengurangan
hasil_3025 = angka1_3025
hasil_3025 -= angka2_3025
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil_3025)

# Assignment perkalian
hasil_3025 = angka1_3025
hasil_3025 *= angka2_3025
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil_3025)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_3025 != 0:
    hasil_3025 = angka1_3025
    hasil_3025 /= angka2_3025
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil_3025)
    # Operator tambahan
    hasil_3025 = angka1_3025
    hasil_3025 //= angka2_3025
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_3025)
    hasil_3025 = angka1_3025
    hasil_3025 %= angka2_3025
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil_3025)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil_3025 = angka1_3025
hasil_3025 **= angka2_3025
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil_3025)
```

**Penjelasan:**
1. Inisialisasi & Konversi Input (int(input())):
Program menerima dua nilai angka numerik dari pengguna melalui fungsi input(). Fungsi int() mengonversi data teks tersebut menjadi integer agar dapat dikenai berbagai operasi penugasan matematika dan menyimpannya ke variabel angka1 serta angka2.

2. Menampilkan Nilai Awal:
Mencetak nilai awal dari angka1 dan angka2 ke layar sebagai titik acuan perhitungan operator assignment.

3. Assignment Biasa (=):
Menggunakan tanda sama dengan = untuk menetapkan nilai dari angka1 secara langsung ke dalam variabel penampung hasil.

4. Assignment Penambahan (+=):
Menggunakan operator += (bentuk ringkas dari hasil = hasil + angka2). Variabel hasil diset ulang ke nilai awal angka1, lalu langsung ditambahkan dengan angka2.

5. Assignment Pengurangan (-=):
Menggunakan operator -= (bentuk ringkas dari hasil = hasil - angka2). Mengurangi nilai variabel hasil dengan angka2.

6. Assignment Perkalian (*=):
Menggunakan operator *= (bentuk ringkas dari hasil = hasil * angka2). Mengalikan nilai variabel hasil dengan angka2.

7. Pengecekan Kondisi Pembagian (if angka2 != 0):
Melakukan validasi pengaman agar program tidak mengalami error (ZeroDivisionError) apabila pengguna memasukkan nilai 0 pada angka kedua.

8. Assignment Pembagian (/=):
Menggunakan operator /= untuk membagi nilai hasil dengan angka2, di mana hasilnya otomatis dikonversi menjadi tipe data pecahan (float).

9. Assignment Pembagian Bulat (//=):
Menggunakan operator //= untuk membagi nilai hasil dengan angka2 dan membulatkan hasilnya ke bawah menjadi bilangan bulat (floor division).

### 2.2.5 Program "Bitwise_2511533025"

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("\n========================================")
print("3. OPERATOR BITWISE")
print("========================================")

angka1_3025 = int(input("Masukkan angka bitwise-1: "))
angka2_3025 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_3025, "| biner =", bin(angka1_3025))
print("angka2 =", angka2_3025, "| biner =", bin(angka2_3025))

# Bitwise AND
hasil_3025 = angka1_3025 & angka2_3025
print("\nBitwise AND (&)")
print(angka1_3025, "&", angka2_3025, "=", hasil_3025)
print("Biner hasil =", bin(hasil_3025))
print("Biner hasil (8 bit) =", format(hasil_3025, "08b"))

# Bitwise OR
hasil_3025 = angka1_3025 | angka2_3025
print("\nBitwise OR (|)")
print(angka1_3025, "|", angka2_3025, "=", hasil_3025)
print("Biner hasil =", bin(hasil_3025))
print("Biner hasil (8 bit) =", format(hasil_3025, "08b"))
# Bitwise XOR
hasil_3025 = angka1_3025 ^ angka2_3025
print("\nBitwise XOR (^)")
print(angka1_3025, "^", angka2_3025, "=", hasil_3025)
print("Biner hasil =", bin(hasil_3025))
print("Biner hasil (8 bit) =", format(hasil_3025, "08b"))

# Bitwise NOT
hasil_3025 = ~angka1_3025
print("\nBitwise NOT (~)")
print("~", angka1_3025, "=", hasil_3025)
print("Biner hasil =", bin(hasil_3025))
print("Biner hasil (8 bit) =", format(hasil_3025, "08b"))
# Bitwise geser kiri
jumlah_geser = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_3025 = angka1_3025 << jumlah_geser
print("\nBitwise geser kiri (<<)")
print(angka1_3025, "<<", jumlah_geser, "=", hasil_3025)
print("Biner hasil =", bin(hasil_3025))
print("Biner hasil (8 bit) =", format(hasil_3025, "08b"))
# Bitwise geser kanan
hasil_3025 = angka1_3025 >> jumlah_geser
print("\nBitwise geser kanan (>>)")
print(angka1_3025, ">>", jumlah_geser, "=", hasil_3025)
print("Biner hasil =", bin(hasil_3025))
print("Biner hasil (8 bit) =", format(hasil_3025, "08b"))
```

**Penjelasan:**
1. Penerimaan Input dan Representasi Bilangan Biner (bin())
Program menerima masukan dua angka integer dari pengguna (angka1_3025 dan angka2_3025). Fungsi bin() digunakan untuk menampilkan representasi biner dari nilai desimal tersebut, yang mengawali string binernya dengan prefiks 0b.

2. Operasi Bitwise AND (&) dan Format 8-Bit (format(..., "08b"))

    a. Ekspresi angka1 & angka2 melakukan perbandingan bit demi bit (bitwise) secara vertikal. Bit hasil akan bernilai 1 jika kedua bit pada posisi yang sama bernilai 1, dan bernilai 0 jika sebaliknya.

    b. Fungsi format(hasil, "08b") digunakan untuk memformat tampilan angka biner ke dalam format tetap 8-bit dengan tambahan angka nol di depan.

3. Operasi Bitwise OR (|) dan XOR (^)

    a. Bitwise OR (|): Menghasilkan bit 1 jika salah satu atau kedua bit pada posisi yang bersesuaian bernilai 1.

    b. Bitwise XOR (^ / Exclusive OR): Menghasilkan bit 1 jika kedua bit yang dibandingkan memiliki nilai yang berbeda (satu bernilai 0 dan satu lagi bernilai 1).

4. Operasi Bitwise NOT (~)
Perintah ~angka1 membalikkan seluruh nilai bit, di mana nilai 0 menjadi 1 dan sebaliknya. Berdasarkan sistem penyimpanan bilangan bulat bertanda dalam komputer, operasi ini menghasilkan nilai -(n + 1).

5. Operasi Pergeseran Bit (Bitwise Shift << dan >>)

    a. Geser Kiri (<<): Menggeser posisi bit ke sebelah kiri sebanyak nilai jumlah_geser, yang secara matematis setara dengan mengalikan angka dengan 2 pangkat jumlah pergeseran.
    
    b. Geser Kanan (>>): Menggeser posisi bit ke sebelah kanan sebanyak nilai jumlah_geser, yang setara dengan membagi angka dengan 2 pangkat jumlah pergeseran secara pembulatan ke bawah.

6. Pencetakan Output Konsol
Fungsi print() menampilkan hasil akhir operasi desimal, representasi biner standar, hingga format biner 8-bit secara terstruktur untuk setiap jenis operator bitwise yang diuji.

### 2.2.6 Program "Lainnya_2511533025"

```python
# Buat file dengan nama lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("========================================")
print("1. OPERATOR KEANGGOTAAN")
print("========================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_3025 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_3025 = [int(angka.strip()) for angka in input_data_3025.split(",")]

nilai_dicari = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_3025 = nilai_dicari in data_3025
print("\nOperator keanggotaan IN")
print(nilai_dicari, "in", data_3025, "=", hasil_3025)

# Operator not in
hasil_3025 = nilai_dicari not in data_3025
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari, "not in", data_3025, "=", hasil_3025)


print("\n========================================")
print("2. OPERATOR IDENTITAS")
print("========================================")

# objek1 menggunakan list dari input pengguna
objek1_3025 = data_3025

# objek2 merujuk pada objek yang sama dengan objek1
objek2_3025 = objek1_3025

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_3025 = data_3025.copy()

print("objek1 =", objek1_3025)
print("objek2 =", objek2_3025)
print("objek3 =", objek3_3025)

# Operator is
hasil_3025 = objek1_3025 is objek2_3025
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil_3025)

# Operator is not
hasil_3025 = objek1_3025 is not objek3_3025
print("\nOperator identitas IS NOT")
print("objek1 is not objek3 =", hasil_3025)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =", objek1_3025 is objek3_3025)
print("objek1 == objek3 =", objek1_3025 == objek3_3025)
```

    b. Operator in: Berfungsi untuk memeriksa apakah suatu nilai tertentu terkandung di dalam list. Jika nilai ditemukan, maka ekspresi menghasilkan nilai Boolean True, dan sebaliknya bernilai False.

    c. Operator not in: Berfungsi untuk memastikan kebalikan dari in, yaitu bernilai True jika nilai yang dicari tidak ada di dalam list tersebut.

2. Operator Identitas (Identity Operators): is dan is not

    a. Perbedaan Alamat Memori: Operator identitas membandingkan apakah dua variabel merujuk pada objek fisik yang sama persis di dalam memori komputer, bukan sekadar membandingkan nilai isinya.

    b. Variabel objek1 dan objek2: objek2 = objek1 membuat kedua variabel menunjuk ke referensi memori yang sama persis, sehingga jika dibandingkan menggunakan objek1 is objek2, hasilnya adalah True.

    c. Penggunaan Metode .copy() (objek3): Perintah data.copy() menduplikasi isi data ke dalam alokasi memori yang baru. Meskipun isi nilai di dalamnya sama, lokasi memorinya berbeda. Oleh karena itu, ekspresi objek1 is not objek3 menghasilkan nilai True.

3. Perbedaan Mendasar Antara Identitas (is) dan Kesetaraan Nilai (==)

    a. Bagian akhir program membandingkan objek1 is objek3 (memeriksa kesamaan alamat memori, menghasilkan False karena memorinya berbeda).

    b. Dibandingkan dengan objek1 == objek3 (memeriksa kesamaan isi atau nilai di dalam objek, menghasilkan True karena elemen angkanya sama persis).

4. Pencetakan Output Konsol
Fungsi print() menampilkan hasil evaluasi boolean dari proses pencarian keanggotaan data di dalam list serta perbandingan alokasi identitas memori objek secara terstruktur.

Sedangkan is (Identitas Objek): Digunakan untuk memeriksa apakah dua variabel merujuk ke objek fisik yang sama persis di dalam memori komputer. Contohnya objek_a is objek_b bernilai False karena keduanya dialokasikan sebagai dua objek terpisah di memori meskipun isi datanya serupa.

5. Operator bitwise digunakan untuk menyimpan kondisi pelanggan dalam bentuk biner. Setiap kondisi memiliki nilai bit, yaitu 0001 untuk member, 0010 untuk belanja minimal, 0100 untuk jumlah barang minimal, dan 1000 untuk promo tersedia. Jika semua kondisi terpenuhi, kode status menjadi 1111. Operator & digunakan untuk memeriksa kondisi tertentu, | untuk menggabungkan kondisi, dan ^ untuk membandingkan perbedaan antara dua kode. Dengan cara ini, program dapat menentukan kondisi dan hak akses pelanggan melalui kode biner.

6. Program dimulai dengan meminta pengguna memasukkan nama, status pelanggan, total belanja, jumlah barang, dan kode promo. Setelah itu, program memeriksa syarat belanja, jumlah barang, status member, dan ketersediaan promo.
Selanjutnya, kondisi tersebut digabungkan menggunakan operator logika untuk menentukan apakah pelanggan mendapatkan diskon atau promo. Jika memenuhi syarat, program menghitung diskon kemudian mengurangkannya dari total belanja untuk mendapatkan total pembayaran. Terakhir, program menampilkan hasil validasi, total pembayaran, status promo, dan hak akses pelanggan.

7. Kondisi yang menghasilkan True: status_member = status == "member" akan bernilai True jika pada saat program berjalan, pengguna mengetikkan status "member". Sedangkan kondisi yang menghasilkan False: belanja_minimal = total_belanja >= 200000 akan bernilai False jika total belanja yang dimasukkan oleh pengguna bernilai di bawah Rp200.000 (misalnya Rp150.000).
