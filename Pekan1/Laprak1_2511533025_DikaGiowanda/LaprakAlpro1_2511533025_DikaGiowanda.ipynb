## 2.1 Dasar Teori
Dasar teori pada praktikum ini membahas konsep dasar yang berkaitan dengan bahasa pemrograman Python, Visual Studio Code (VS Code), dan GitHub. Ketiga konsep tersebut merupakan bagian penting dalam proses pengembangan program, mulai dari penulisan kode hingga pengelolaan dan penyimpanan kode secara daring. Contoh:

1. Python, Python merupakan bahasa pemrograman tingkat tinggi yang memiliki sintaks sederhana dan mudah dipahami. Python banyak digunakan untuk berbagai kebutuhan, seperti pengembangan aplikasi, pengolahan data, kecerdasan buatan, dan otomasi. Dalam praktikum ini, Python digunakan untuk mempelajari dasar penulisan dan eksekusi program. Beberapa elemen penting yang digunkan antara lain, Komentar(#): digunakan untuk memberi catatan atau dokumentasi pada kode tanpa dieksekusi oleh program, Output (print()): Fungsi standar di Python untuk menampilkan data atau teks ke layar, Karakter Escape: Karakter khusus yang diawali dengan tanda garis miring terbalik (\), dan Penggabungan String: Proses menggabungkan dua atau lebih teks (string) menjadi satu kesatuan dengan menggunakan operator penjumlahan (+).
2. ⁠Visual Studio Code (VScode), Visual Studio Code adalah aplikasi code editor yang digunakan untuk menulis dan mengembangkan program. VS Code menyediakan berbagai fitur yang mendukung pemrograman, seperti syntax highlighting, terminal, debugging, serta extension. Dalam praktikum ini, VS Code digunakan sebagai tempat untuk menulis dan menjalankan kode Python.

3. Git dan Github, Git adalah sistem pengontrol versi terdistribusi yang digunakan untuk mencatat setiap perubahan pada berkas kode proyek. GitHub adalah platform berbasis cloud yang memanfaatkan Git untuk menyimpan repositori proyek, memfasilitasi pengarsipan kode, serta memungkinkan kolaborasi antar pengembang.

4. Koneksi VS Code dengan GitHub, VS Code dapat dihubungkan dengan GitHub sehingga proses pengelolaan kode menjadi lebih mudah. Kode yang dibuat melalui VS Code dapat di-commit dan di-push ke repositori GitHub. Dengan demikian, setiap perubahan pada program dapat terdokumentasi dan tersimpan secara daring.

5. Repository dan Version Control, Repository merupakan tempat penyimpanan proyek yang berisi file dan riwayat perubahan kode. Version control memungkinkan pengguna mengetahui perubahan yang telah dilakukan pada suatu proyek serta membantu mengembalikan kode ke versi sebelumnya apabila terjadi kesalahan.

```python
# Program Tugas1_2511533025.py
"""
Praktikum Algoritma dan Pemrograman
Hitung Nilai Akhir Mahasiswa
"""
# Deklarasi data dan variabel
nama_mahasiswa_3025 = "Dika Giowanda"
nim_3025 = "2511533025"
nilai_tugas_3025 = 100
nilai_uts_3025 = 95
nilai_uas_3025 = 100

# Proses perhitungan nilai akhir secara otomatis
# (Tugas 20%, UTS 30%, UAS 50%)
nilai_akhir_3025 = (nilai_tugas_3025 * 0.2) + (nilai_uts_3025 * 0.3) + (nilai_uas_3025 * 0.5)

# Tampilan Output
print("====== KARTU HASIL STUDI ======")
print("Nama Mahasiswa\t:", nama_mahasiswa_3025)
print("NIM\t\t:", nim_3025)
print("-------------------------------")
print("Nilai Tugas\t:", nilai_tugas_3025)
print("Nilai UTS\t:", nilai_uts_3025)
print("Nilai UAS\t:", nilai_uas_3025)
print("-------------------------------")
print("Nilai Akhir\t:", nilai_akhir_3025)
print("===============================")
print("Laporan ini" + " dibuat oleh\n" 
      + "Asisten Laboratorium Alpro.")
```

**Penjelasan:**
1. ⁠Komentar Multi-baris (""" ... """)

    a. Fungsi: Menjelaskan identitas program, seperti judul program dan topik praktikum.
    
    b. Cara Kerja: Karakter # digunakan untuk komentar satu baris, sedangkan tanda petik tiga (""") digunakan untuk membuat komentar beberapa baris. Teks di dalam bagian ini diabaikan oleh Python dan tidak akan muncul pada output.

2. Deklarasi Variabel dan Inisialisasi Data

    a. Fungsi: Menyimpan data awal ke dalam memori komputer berupa variabel agar dapat digunakan pada proses selanjutnya.
    
    b. Cara Kerja: nama_mahasiswa dan npm menyimpan data bertipe String (teks) yang ditandai dengan tanda petik dua (""). nilai_tugas, nilai_uts, dan nilai_uas menyimpan data bertipe Integer (bilangan bulat) tanpa tanda petik, karena nilainya akan dihitung menggunakan operasi matematika.

3. Proses Perhitungan (Operator Aritmatika)

    a. Fungsi: Menghitung nilai akhir mahasiswa secara otomatis berdasarkan persentase bobot nilai.
    
    b. Cara Kerja: Operator (*) digunakan untuk operasi perkalian, dan operator (+) untuk penjumlahan. Tanda kurung () memastikan kalkulasi perkalian bobot dikerjakan terlebih dahulu. Rumus menghitung: 20% dari nilai tugas (0.2), 30% dari nilai UTS (0.3), dan 50% dari nilai UAS (0.5). Hasil perhitungan disimpan di dalam variabel baru bernama nilai_akhir (bertipe data Float / desimal).

4. Tampilan Output dan Formatting (print)

    a. Fungsi: Menampilkan informasi hasil perhitungan ke layar terminal.

    b. Cara Kerja: print() digunakan untuk mencetak teks maupun isi dari variabel. Karakter \t (tabulation) berfungsi memberikan jarak spasi horizontal (seperti tombol Tab) agar posisi tanda titik dua (:) tercetak rapi dan sejajar ke bawah. Variabel NIM\t\t: memakai dua \t karena kata "NIM" lebih pendek dibanding kata "Nama Mahasiswa", sehingga butuh jarak ekstra agar titik duanya sejajar.

5. Penggabungan String (Concatenation) dan Escape Sequence

    a. Fungsi: Menampilkan teks penutup program dengan format khusus.

    b. Cara Kerja: Operator (+) digunakan untuk menyambungkan dua atau lebih variabel/teks. Karakter \n (newline) berfungsi membuat baris baru (pindah baris ke bawah) di tengah-tengah kalimat.

Tahapan Commit dan Push file ke akun Github:

1. Klik tombol biru "Publish to GitHub".
2. Jika ada pop up konfirmasi atau instruksi untuk login/authorize ke akun GitHub milikmu, klik Allow dan ikuti proses login-nya di browser.
3. Pilih opsi opsi repository:
Publish to GitHub private repository (jika ingin repository hanya bisa dilihat olehmu/pengajar)
Publish to GitHub public repository (jika ingin bisa diakses siapa saja)
4. Centang file tugas dan laporan yang ingin diunggah, lalu klik OK.
5. VS Code akan otomatis melakukan Initialize, Commit pertama, dan Push ke GitHub.
6. Jika nanti kamu menambahkan atau mengedit file tugas/laporan lagi di VS Code, buka kembali tab Source Control (ikon cabang di sebelah kiri).
7. Ketik pesan perubahan pada kolom Message (contoh: "Menambahkan file laporan tugas").
8. Klik tombol Commit.
9. Klik tombol Sync Changes (atau Publish Branch / Push) yang muncul untuk mengunggah perubahan ke GitHub.
