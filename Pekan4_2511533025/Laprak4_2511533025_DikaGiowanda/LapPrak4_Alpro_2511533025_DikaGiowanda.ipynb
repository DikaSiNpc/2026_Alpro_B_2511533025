## 2.1 Dasar Teori
Statement Conditional (percabangan atau pengondisian) merupakan fitur dalam bahasa pemrograman yang memungkinkan sebuah program mengambil keputusan dengan mengeksekusi blok kode tertentu berdasarkan kondisi logika bernilai Boolean (True atau False). Pada bahasa pemrograman Python, pengondisian diatur menggunakan indentasi (spasi/tab) untuk menentukan cakupan blok kode.
Berikut adalah poin-poin konsep dan teori dasar yang mendukung materi Statement Conditional pada Python:
1. Pengertian Conditional Statement dan Evaluasi Boolean

    a. Conditional statement berfungsi mengontrol alur eksekusi program (flow control).

    b. Kondisi dievaluasi menggunakan ekspresi logika atau perbandingan yang menghasilkan nilai True atau False. Jika kondisi terpenuhi (True), blok kode di dalamnya akan dijalankan.

2. Operator Perbandingan dan Operator Logika

    a. Operator Perbandingan: Digunakan untuk membandingkan dua nilai, seperti == (sama dengan), != (tidak sama dengan), < (kurang dari), > (lebih dari), <= (kurang dari atau sama dengan), dan >= (lebih dari atau sama dengan).

    b. Operator Logika: Digunakan untuk menggabungkan beberapa kondisi logika, yaitu and (semua kondisi harus True), or (salah satu kondisi True), dan not (membalikkan nilai kebenaran).

3. Struktur Percabangan Tunggal (if)

    a. Merupakan bentuk kondisional paling sederhana.

    b. Blok kode di dalam perintah if hanya akan dieksekusi jika kondisi yang didefinisikan bernilai True. Jika bernilai False, program akan langsung melompati blok kode tersebut.

4. Struktur Percabangan Dua Kondisi (if-else)

    a. Digunakan untuk menangani dua kemungkinan alur eksekusi.

    b. Jika kondisi pada if bernilai True, maka blok if dijalankan. Namun jika kondisi bernilai False, program secara otomatis akan mengeksekusi blok kode alternatif pada else.

5. Struktur Percabangan Banyak Kondisi (if-elif-else)

    a. Digunakan ketika terdapat lebih dari dua kondisi yang perlu diperiksa secara berurutan.

    b. Keyword elif (else if) memeriksa kondisi berikutnya secara sekuensial jika kondisi sebelumnya bernilai False. Blok else opsional diletakkan di akhir sebagai pilihan terakhir jika semua kondisi if maupun elif tidak terpenuhi.

6. Struktur Percabangan Independen/Majemuk (Multi-if)

    a. Berbeda dengan if-elif-else, struktur Multi-if terdiri dari beberapa perintah if terpisah yang berdiri sendiri.

    b. Setiap kondisi if akan dievaluasi satu per satu secara independen, sehingga memungkinkan lebih dari satu blok kode dieksekusi apabila beberapa kondisi sama-sama bernilai True.

7. Struktur Percabangan Bersarang (Nested if)

    a. Merupakan teknik menempatkan struktur if di dalam struktur if lainnya.

    b. Digunakan ketika suatu kondisi perlu diperiksa hanya setelah kondisi utama/sebelumnya telah terpenuhi (True).

8. Pencocokan Pola Struktural (Structural Pattern Matching / match-case)

    a. Diperkenalkan mulai versi Python 3.10 sebagai alternatif yang lebih rapi dan efisien dari struktur if-elif-else yang panjang.

    b. Menggunakan sintaks match untuk menentukan variabel yang diperiksa dan case untuk mencocokkan nilainya, serta case _: sebagai penanganan kondisi default (setara dengan else).

### 2.2.1 Program "if1_2511533025"

```python
# Buat file dengan nama if1_NIM.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ifk_1234
# Program ini menggunakan fungsi input()

ipk_3025 = float(input("Input IPK Anda = "))

if ipk_3025 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_3025))

print("Program Selesai")
```

    d. else: Merupakan blok alternatif terakhir jika seluruh kondisi if maupun elif di atas bernilai False. Blok ini akan mencetak: "Anda Belum Cukup Umur dan tidak boleh bawa motor".

### 2.2.5 Program "multi_if2_2511533025"

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_3025 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_3025 = input("Apakah Anda memiliki member? (y/t): ").strip().lower()
is_member_3025 = input_member_3025 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3025 = input("Apakah kode promo anda valid? (y/t): ").strip().lower()
kode_promo_valid_3025 = input_promo_3025 in ["y", "ya"]

total_diskon_persen_3025 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3025 > 1000000:
    total_diskon_persen_3025 += 10  # Diskon belanja besar

if is_member_3025:
    total_diskon_persen_3025 += 5   # Diskon member
if kode_promo_valid_3025:
    total_diskon_persen_3025 += 15  # Diskon voucher

# Menhitung nominal diskon dan total bayar
nominal_diskon_3025 = total_belanja_3025 * (total_diskon_persen_3025 / 100)
total_bayar_3025 = total_belanja_3025 - nominal_diskon_3025

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_3025}% (Rp {nominal_diskon_3025:,.0f})")
print(f"Total Bayar   : Rp {total_bayar_3025:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_3025}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan :**
1. Bagian Pertama (Input Data dan Pembersihan Teks):

    a. total_belanja = float(input(...)): Berfungsi untuk menerima masukan nilai total belanja dari pengguna dalam bentuk angka desimal (float).

    b. input_member dan input_promo: Menerima masukan teks status member dan kode promo. Metode .strip() digunakan untuk menghapus spasi di awal/akhir teks, dan .lower() untuk mengubah huruf menjadi huruf kecil (lowercase) agar input pengguna lebih fleksibel (misal mengantisipasi ketikan "Y" atau "YA").

    c. is_member dan kode_promo_valid: Mengonversi input teks tersebut menjadi nilai kebenaran (Boolean) dengan memeriksa apakah teks berada di dalam daftar ["y", "ya"].

2. Bagian Kedua (Struktur Multi-If Terpisah dan Akumulasi):

    a. Berbeda dengan struktur if-elif-else yang saling mengikat, program ini menggunakan struktur Multi-IF terpisah di mana setiap blok if berdiri sendiri dan dievaluasi satu per satu secara independen.

    b. total_diskon_persen += ...: Operator penambahan bersyarat ini memungkinkan persentase diskon ditumpuk atau diakumulasikan apabila pengguna memenuhi lebih dari satu syarat sekaligus:

    c. if total_belanja > 1000000: Menambahkan diskon 10% jika total belanja melebihi Rp 1.000.000.

    d. if is_member: Menambahkan diskon 5% jika status member bernilai True.

    e.if kode_promo_valid: Menambahkan diskon 15% jika kode promo valid bernilai True.

    f. Contoh: Jika belanja di atas 1 juta, member, dan promo valid, maka total diskon akan terakumulasi menjadi 10%+5%+15%=30%.

3. Bagian Ketiga (Perhitungan Nominal dan Total Bayar):

    a. nominal_diskon = total_belanja * (total_diskon_persen / 100): Menghitung besar potongan harga dalam bentuk nominal uang rupiah berdasarkan total persentase diskon yang terkumpul.

    b. total_bayar = total_belanja - nominal_diskon: Menghitung jumlah bersih uang yang harus dibayarkan setelah total belanja dikurangi nominal diskon.

4. Bagian Keempat (Pencetakan Output):

    a. Menggunakan format string (f"...") untuk menampilkan rincian pembayaran secara rapi ke layar.

    b. Format :,.0f pada variabel angka berfungsi untuk menampilkan format mata uang dengan pemisah ribuan koma/titik dan tanpa angka desimal di belakang koma, sehingga laporan rincian biaya lebih mudah dibaca.

### 2.2.6 Program "Nilai_2511533025"

### 2.2.7 Program "match_case1_2511533025"

```python
# Buat file dengan nama match_case1.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# program konversi angka menjadi nama bulan

bulan_3025 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_3025:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan :**
1. Bagian Pertama 

    a. (bulan = int(input("Masukkan angka bulan (1 – 12): "))) Berfungsi untuk menerima masukan angka bulan dari pengguna secara interaktif melalui papan ketik.

    b. Fungsi input() mengambil data dalam bentuk teks (string), kemudian fungsi int() mengonversinya menjadi bilangan bulat (integer) agar dapat dicocokkan dengan pola angka pada struktur match-case, lalu disimpan ke dalam variabel bulan.

2. Bagian Kedua (Struktur Percabangan Pencocokan Pola / Match-Case):

    a. Berfungsi sebagai struktur kontrol keputusan modern yang diperkenalkan pada Python versi 3.10 ke atas untuk menangani percabangan banyak kondisi secara lebih rapi dan terstruktur:

    b. match bulan: Menentukan variabel utama yang nilai atau polanya akan diperiksa.

    c. Blok case 1: hingga case 12:: Masing-masing cabang case memeriksa apakah nilai variabel bulan cocok dengan angka tertentu (1 sampai 12). Jika cocok, program mengeksekusi perintah print() di dalam blok tersebut untuk mencetak nama bulan yang bersesuaian (misalnya case 5: mencetak "Mei").

    d. case _:: Berfungsi sebagai wildcard atau kondisi default. Blok ini akan dieksekusi apabila nilai yang dimasukkan oleh pengguna tidak ada yang cocok dengan angka 1 sampai 12, dan mencetak pesan "Angka tidak valid".

3. Bagian Analisis Output:

    Hasil keluaran program menunjukkan teks Mei. Hal ini berarti angka bulan yang dimasukkan oleh pengguna cocok dengan pola pada case 5:, sehingga blok kode di bawahnya dieksekusi dan menampilkan nama bulan "Mei" ke layar.

```python
# =========================================================
# SISTEM LOKET TERPADU & AUDIT TRANSAKSI EKSPEDISI WAHANA
# Studi Kasus: Alpro Adventure Park
# Materi: Struktur Percabangan (Pekan 4)
# =========================================================

print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# ---------------------------------------------------------
# 1. Input Data Pengunjung & String Handling
# ---------------------------------------------------------
nama_3025 = input("Masukkan Nama Pengunjung\t   : ")
umur_3025 = int(input("Input umur anda\t\t\t   : "))
sim_3025 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0].strip().lower()

# ---------------------------------------------------------
# 2. Pemilihan Wahana menggunakan match-case (1-5 & default _)
# ---------------------------------------------------------
print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba       (Rp 50,000)")
print("  2. Arung Jeram        (Rp 75,000)")
print("  3. Motor ATV Ekstrim  (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP     (Rp 220,000)")

pilihan_paket_3025 = int(input("Masukkan nomor paket anda (1-5)\t: "))

match pilihan_paket_3025:
    case 1:
        nama_paket_3025 = "Wahana Safari Rimba"
        harga_satuan_3025 = 50000
    case 2:
        nama_paket_3025 = "Wahana Arung Jeram"
        harga_satuan_3025 = 75000
    case 3:
        nama_paket_3025 = "Wahana Motor ATV Ekstrim"
        harga_satuan_3025 = 120000
    case 4:
        nama_paket_3025 = "Wahana Roller Coaster Kilat"
        harga_satuan_3025 = 100000
    case 5:
        nama_paket_3025 = "Wahana All-Access VIP"
        harga_satuan_3025 = 220000
    case _:
        print("Paket wahana tidak valid!")
        exit()

# ---------------------------------------------------------
# 1b. Input Jumlah Tiket + validasi if tunggal
# ---------------------------------------------------------
jumlah_tiket_3025 = int(input("Masukkan jumlah tiket anda\t: "))

if jumlah_tiket_3025 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")

# ---------------------------------------------------------
# 3. Validasi Izin Kendali Wahana (if-elif-else + operator logika)
#    Tidak menggunakan nested-if: syarat paket 3 digabung
#    langsung ke setiap cabang kondisi menggunakan 'and'
# ---------------------------------------------------------
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if pilihan_paket_3025 == 3 and umur_3025 >= 17 and sim_3025 == 'y':
    print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
elif pilihan_paket_3025 == 3 and umur_3025 >= 17 and sim_3025 != 'y':
    print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV "
          "(wajib didampingi instruktur).")
elif pilihan_paket_3025 == 3 and umur_3025 < 17 and sim_3025 == 'y':
    print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
elif pilihan_paket_3025 == 3:
    print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
elif umur_3025 >= 10:
    print(f"Status Akses: Anda cukup umur untuk menaiki {nama_paket_3025}.")
else:
    print(f"Status Akses: Anda belum cukup umur untuk menaiki {nama_paket_3025}.")

# ---------------------------------------------------------
# Input tambahan: status member & kode promo
# ---------------------------------------------------------
is_member_3025 = input("Apakah anda memiliki member? (y/t) : ").strip().lower()
kode_promo_valid_3025 = input("Apakah kode promo anda valid? (y/t): ").strip().lower()

# ---------------------------------------------------------
# 4. Akumulasi Diskon Bertingkat (Multi-IF terpisah)
# ---------------------------------------------------------
subtotal_3025 = harga_satuan_3025 * jumlah_tiket_3025
total_diskon_persen_3025 = 0

if subtotal_3025 >= 200000:
    total_diskon_persen_3025 += 10          # Diskon Belanja Besar

if is_member_3025 in ['y', 'ya']:
    total_diskon_persen_3025 += 5           # Diskon Member

if kode_promo_valid_3025 in ['y', 'ya']:
    total_diskon_persen_3025 += 15          # Diskon Voucher Promo

if jumlah_tiket_3025 >= 5:
    total_diskon_persen_3025 += 5           # Diskon Tambahan Rombongan

# ---------------------------------------------------------
# 5. Evaluasi Kelulusan Audit (if-else) + Rincian Pembayaran
# ---------------------------------------------------------
nominal_diskon_3025 = subtotal_3025 * (total_diskon_persen_3025 / 100)
total_bayar_3025 = subtotal_3025 - nominal_diskon_3025

print("\n--- Rincian Pembayaran ---")
print(f"Subtotal Belanja : Rp {subtotal_3025:,.0f}")
print(f"Total Diskon\t : {total_diskon_persen_3025}% (Rp {nominal_diskon_3025:,.0f})")
print(f"Total Bayar\t : Rp {total_bayar_3025:,.0f}")

if total_bayar_3025 > 300000:
    print("Catatan Layanan  : Selamat! Anda berhak mendapatkan Souvenir Gratis.")
else:
    print("Catatan Layanan  : Terima kasih telah berkunjung.")

print("Program Selesai")
```

**Penjelasan :**
1. Bagian Input Data Pengunjung & Handling String, berfungsi untuk menerima input awal berupa nama, umur (dikonversi ke tipe data int), serta kepemilikan SIM C. String input SIM C diproses menggunakan indeks [0] untuk mengambil karakter pertama, .strip() untuk menghapus spasi berlebih, dan .lower() untuk mengubah huruf menjadi kecil agar konsisten.

2. Bagian Pemilihan Wahana (match-case), berfungsi menentukan jenis paket dan harga tiket satuan berdasarkan pilihan angka (1-5). Jika pengguna menginput angka di luar batas 1–5, variabel akan ditangkap oleh default case (_), menampilkan pesan kesalahan, dan menghentikan eksekusi program menggunakan fungsi exit().

3. Bagian Input Jumlah Tiket & Validasi Tunggal, berfungsi mengambil input kuota tiket dari pengguna dan menggunakan struktur if tunggal untuk memeriksa apakah jumlah tiket yang diinput bernilai kurang dari atau sama dengan nol (≤0). Jika iya, sistem akan memberikan peringatan kuota tidak valid.

4. Bagian Validasi Izin Kelayakan Pengendara (if-elif-else dengan Operator Logika), berfungsi menentukan hak akses dan tingkat keselamatan pengunjung saat menaiki/mengendarai wahana tanpa menggunakan nested-if. Kondisi diperiksa secara runtut menggunakan gabungan operator logika and. Wahana paket 3 (ATV Ekstrim) diverifikasi berdasarkan kombinasi kriteria umur dan kepemilikan SIM C, sedangkan paket wahana lainnya cukup diverifikasi dengan kriteria batas umur minimal 10 tahun.

5. Bagian Input Status Tambahan (Member & Promo), berfungsi mengumpulkan informasi mengenai status keanggotaan (member) dan keabsahan kode promo pengguna yang akan digunakan pada tahapan kalkulasi diskon.

6. Bagian Akumulasi Diskon Bertingkat (Multi-if Terpisah), berfungsi untuk menghitung akumulasi total persentase diskon yang berhak didapatkan pelanggan. Menggunakan struktur beberapa perintah if terpisah (bukan elif) agar setiap kondisi diskon dapat dievaluasi secara independen dan nilainya dijumlahkan secara kumulatif (total_diskon_persen += ...). Kriteria diskon mencakup:
Diskon Belanja Besar (+10%) jika subtotal ≥Rp 200.000.
Diskon Member (+5%) jika terdaftar sebagai member.
Diskon Kode Promo (+15%) jika kode promo valid.
Diskon Rombongan (+5%) jika jumlah tiket ≥5.

7. Bagian Perhitungan Pembayaran & Evaluasi Audit (if-else), berfungsi menghitung nominal diskon dalam rupiah, subtotal setelah diskon (total bayar), dan mencetak rincian struk pembayaran. Selanjutnya, terdapat percabangan if-else untuk mengevaluasi apakah total bayar melebihi Rp 300.000 untuk memberikan souvenir gratis sebelum ditutup dengan pernyataan bahwa program telah selesai.
